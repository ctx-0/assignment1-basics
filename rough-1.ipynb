Unicode is a text encoding standard that maps characters to *integer code points*.

*Notation*  
> code point 115 (typically notated as U+0073, where U+ is a conventional prefix and 0073 is 115 in hexadecimal)

*Encoding*  
> Unicode encoding, which converts a Unicode character *into a sequence of bytes*  


>  UTF-8, UTF-16, and UTF-32, with UTF-8 being the dominant encoding for the 
Internet (more than 98% of all webpages)
